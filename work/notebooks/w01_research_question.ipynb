Cost of a Wrong Call:

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

We load the raw search dataset to establish three core performance metrics: total unique page volume, average SERP ranking position, and domain-wide average Click-Through Rate.

```python
import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url).fillna(0)

# Calculate 3 real metrics
total_pages = len(df)
mean_impressions = df["impressions_90d"].mean()
mean_ctr = df["ctr"].mean() * 100

print(f"1. Total Unique Pages Evaluated: {total_pages:,}")
print(f"2. Average 90-Day Impressions per Page: {mean_impressions:,.1f}")
print(f"3. Domain Average Click-Through Rate: {mean_ctr:.2f}%")
```
